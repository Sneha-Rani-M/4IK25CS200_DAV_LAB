 DATASET LOADING

```python
df = pd.read_csv('/content/student-mat.csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

Step 2: Understand the dataset

```python
# Display dataset information
df.info()
```

```python
# Display statistical summary
df.describe()
```

Step 3: Check missing values and duplicate records

```python
# Check missing values in each column
print("Missing values:")
print(df.isnull().sum())

# Check total missing values
print("\nTotal missing values:", df.isnull().sum().sum())

# Check duplicate records
print("\nTotal duplicate records:", df.duplicated().sum())
```

Step 4: Data cleaning

```python
# Remove duplicate records
df = df.drop_duplicates()

# Fill missing numerical values with their column mean
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].mean())

# Fill missing categorical values with their mode
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Data cleaning completed!")
print("Dataset shape:", df.shape)

# Verify missing values
print("Total missing values:", df.isnull().sum().sum())
```

Step 5: Exploratory Data Analysis (EDA)

```python
# Basic analysis of student performance

print("Average Final Grade:", df['G3'].mean())
print("Highest Final Grade:", df['G3'].max())
print("Lowest Final Grade:", df['G3'].min())
print("Average Age:", df['age'].mean())
print("Total Students:", len(df))
```

Step 6: Visualize the distribution of student grades

```python
plt.figure(figsize=(8, 5))

sns.histplot(df['G3'], bins=10, kde=True)

plt.title("Distribution of Students' Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

Step 7: Compare performance by gender

```python
plt.figure(figsize=(7, 5))

sns.barplot(data=df, x='sex', y='G3')

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

plt.show()
```

Step 8: Analyse study time and student performance

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='studytime', y='G3')

plt.title("Study Time vs Final Grade")
plt.xlabel("Weekly Study Time")
plt.ylabel("Average Final Grade")

plt.show()
```

Step 9: Analyse absences and student performance

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x='absences', y='G3')

plt.title("Student Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")

plt.show()
```

Step 10: Correlation analysis

```python
# Select numerical columns
numeric_df = df.select_dtypes(include=np.number)

# Calculate correlation
correlation = numeric_df.corr()

# Display correlation heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title("Correlation Heatmap of Student Performance")

plt.show()
```

Step 11: Compare previous grades with final grades

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x='G1', y='G3', label='G1')
sns.scatterplot(data=df, x='G2', y='G3', label='G2')

plt.title("Previous Grades vs Final Grade")
plt.xlabel("Previous Grades")
plt.ylabel("Final Grade (G3)")

plt.legend()
plt.show()
```

Step 12: Key Findings and Insights

```python
# Key findings

print("KEY FINDINGS")
print("--------------------------")

print("1. Total Students:", len(df))

print("2. Average Final Grade:", round(df['G3'].mean(), 2))

print("3. Highest Final Grade:", df['G3'].max())

print("4. Lowest Final Grade:", df['G3'].min())

print("5. Average Study Time:", round(df['studytime'].mean(), 2))

print("6. Average Absences:", round(df['absences'].mean(), 2))

print("7. Average Grade of Female Students:",
      round(df[df['sex'] == 'F']['G3'].mean(), 2))

print("8. Average Grade of Male Students:",
      round(df[df['sex'] == 'M']['G3'].mean(), 2))
```
